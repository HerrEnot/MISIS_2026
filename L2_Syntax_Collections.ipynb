# 1. Git за десять минут

Начнём с вопроса, который на первый взгляд к программированию отношения не имеет. Как выглядит папка с вашей курсовой? Честно. Я подожду.

Скорее всего, вот так:

- `Курсовая.docx`
- `Курсовая_правки.docx`
- `Курсовая_после_замечаний_научрука.docx`
- `Курсовая_ФИНАЛ.docx`
- `Курсовая_ФИНАЛ_v2_ТОЧНО_ПОСЛЕДНИЙ.docx`
- `Курсовая_ФИНАЛ_v2_ТОЧНО_ПОСЛЕДНИЙ (1).docx`

Это хаос. И это хаос, который однажды вас убьёт. Вы удалили важный абзац в `v2` и хотите вернуть его из версии `после_замечаний` — а какая из шести это была? Одногруппник правил свою копию, вы — свою, и теперь вам надо склеить две версии руками в пятницу в 23:50. Знакомо?

Проблема старая как мир, и решение у неё тоже старое: **системы контроля версий** (Version Control Systems, VCS). Самая популярная из них сегодня, с огромным отрывом, — **Git**.



*Кому нужны Git и SVN, когда есть файлы «Final», «Final 2» и «FINAL FOR REAL THIS TIME». Идеально. Ровно до того дня, когда вы откроете не ту.*

## Аналогия: Git — это сохранения в игре

Забудьте умные определения. Представьте, что ваш проект — это длинная сложная игра.

- **Репозиторий** (repository) — папка с вашей игрой. С вашим проектом.
- **Коммит** (commit) — точка сохранения. Дошли до важного места — написали работающую функцию, закончили главу — и нажали «Сохранить игру». Каждому сохранению даёте осмысленное имя: «Победил первого босса», «Добавил расчёт среднего чека». Если дальше всё пойдёт прахом, вы всегда загрузитесь с этого сохранения.
- **Ветка** (branch) — альтернативная временная линия. Хотите пройти уровень с новым мечом, но не потерять старое сохранение? Создаёте ветку, экспериментируете сколько угодно. Получилось — сливаете с основной линией. Не получилось — удаляете, и никто никогда не узнает, какой ерундой вы там занимались.

А **Git** — это движок игры, который всеми этими сохранениями и временными линиями управляет.

Сохраняться, кстати, надо часто. Не раз в неделю перед дедлайном — а каждый раз, когда что-то заработало.



*В играх сохраняемся каждые пять минут. В коде — один коммит `fix tests`: +647 и −1049 строк разом. Коммит — это сохранение, а не надгробие.*

## Git и GitHub — это разные вещи

Самый частый источник путаницы, проясним раз и навсегда.

**Git** — это программа. Живёт у вас на компьютере, работает локально, следит за изменениями файлов, делает коммиты и ветки. Аналогия: Microsoft Word — программа, которая стоит у вас на ноутбуке.

**GitHub** — это сайт. Облако, куда вы отправляете свои репозитории, чтобы хранить их, показывать и работать над ними вместе. Аналогия: Google Docs или Яндекс Диск. Кроме GitHub есть GitLab, Gitflic, Bitbucket — принцип тот же.

| | Git | GitHub |
| :--- | :--- | :--- |
| **Что это** | Программа | Сайт, сервис |
| **Где работает** | На вашем компьютере | В интернете |
| **Зачем** | Хранить историю изменений | Хранить репозитории в облаке, работать вместе |

Итого: Git ведёт историю проекта у вас на компьютере, а потом вы отправляете эту историю на GitHub — чтобы не потерять, когда умрёт ноутбук, и чтобы я мог её проверить.

**Зачем это аналитику, а не только программисту?** Анализ данных — это исследование: вы пробуете гипотезы, меняете код, перестраиваете графики. Git позволяет вернуться к той версии, которая дала интересный результат, вместо того чтобы хранить десять копий тетрадки с суффиксами `_new`, `_new2`, `_рабочая`. А в команде — а вы будете работать в команде — без Git вы просто не сможете работать.



*Merge-конфликт: двое поменяли одну и ту же строку. Git честно покажет, кто виноват. Спойлер: это были вы.*

### Шаг 2. Коммит и публикация — мышкой в VS Code

Дальше никаких страшных команд:

1. Откройте папку проекта в VS Code и панель **Source Control**.
2. Там вы увидите список изменённых файлов. Наведите на файл и нажмите `+` — это значит «я хочу положить этот файл в следующее сохранение».
3. Сверху в поле **Message** напишите, что вы сделали. Не «asdf» и не «fix», а «ДЗ1: проверка високосного года». Нажмите **Commit**. Поздравляю, вы сохранили игру.
4. Первый раз VS Code предложит кнопку **Publish Branch** (или **Publish to GitHub**). Нажимаете, входите в свой аккаунт GitHub, выбираете **Public repository**. Всё — ваш проект в облаке.
5. Дальше каждый раз: изменили → `+` → сообщение → **Commit** → **Sync Changes** (отправить на GitHub).

Репозитории у нас **публичные**. Да, это значит, что ваш код может посмотреть кто угодно, включая одногруппников. Списывайте на здоровье — на зачёте я попрошу поправить ваш код у меня на глазах, и всё встанет на свои места.

*🎤 Обязательно покажите это вживую на своём ноутбуке: правка в файле → Source Control → Commit → Sync. После десяти минут текста им надо увидеть, что это три клика, а не магия.*

## Главное правило Git: не коммитьте секреты

Пароли, токены, ключи от API — **никогда** в репозиторий. Всё, что попало в коммит, остаётся в истории навсегда. Удалить файл следующим коммитом — бесполезно: старое сохранение-то никуда не делось. Через три лекции у вас будет токен Telegram-бота. Положите его в публичный репозиторий — и рано или поздно ваш бот начнёт жить своей жизнью: рассылать чужой спам от вашего имени. И «рано» тут — не фигура речи: ключи от облака, как вы сейчас узнаете, уводят за считаные минуты. Как это происходит — в экстра-блоке ниже.

Давайте я расскажу, как это работает на самом деле, чтобы вы прониклись.

Публичный GitHub круглосуточно сканируют боты. Не люди — программы, которые читают каждый новый коммит и ищут в нём строки, похожие на ключи. Исследователи Palo Alto Networks (группа Unit 42) в 2023 году специально выкладывали ключи от облака Amazon на GitHub и засекали время. **Меньше чем за пять минут** ключ находили, забирали и запускали на облачном аккаунте майнинг криптовалюты. За счёт владельца аккаунта. В похожем эксперименте 2024 года (Clutch Security) ключ с GitHub увели **примерно за минуту**, а ключ, забытый в Docker-образе на Docker Hub, — **меньше чем за сорок секунд**. Вы ещё не успели отойти от компьютера, а счёт за облако уже тикает на тысячи долларов.

Масштаб: по отчёту GitGuardian за 2025 год в публичных коммитах GitHub нашли **28,6 миллиона** новых секретов. И самое смешное — из секретов, которые утекли ещё в 2022 году и тогда были рабочими, **больше 64 % продолжали работать** в начале 2026-го. Люди не просто роняют ключи — они их потом даже не меняют.

**Что делать, если вы всё-таки закоммитили секрет:**
1. **Первым делом — отозвать ключ.** Пойти в BotFather, в кабинет облака, куда угодно — и сказать «этот ключ больше недействителен, выдайте новый». Это единственное, что реально спасает.
2. Только потом — чистить историю (инструмент `git filter-repo`). Чистка истории без замены ключа — это как вытереть отпечатки пальцев, оставив ключ в замке.

**Как не закоммитить секрет вообще.** Секреты живут в отдельном файле `.env`, а этот файл записан в `.gitignore` — **до** первого коммита. Вот как это выглядит в терминале — в macOS, Linux или в **Git Bash** (он ставится на Windows вместе с Git):
```bash
echo 'TG_TOKEN=123456:ABC-секрет' > .env
echo '.env' >> .gitignore          # ДО первого git add!
git add . && git status            # .env в списке нет — и хорошо
git log -p --all -S 'ABC-секрет'   # так ищут секрет во всей истории — и вы, и боты
```

Последняя команда — это ровно то, что делают боты: ищут строку по всей истории, включая «удалённые» файлы.

**В PowerShell на Windows первые две строки так не пишите:** `echo … >>` там сохраняет файл в кодировке, которую Git не понимает, — `.gitignore` молча перестанет работать, и `.env` уедет в коммит. На Windows проще создать `.env` и `.gitignore` в VS Code (**New File**) и вписать строки руками. А проверка в любом терминале одна: `git status` — `.env` в списке быть не должно.

GitHub, к слову, с февраля 2024 года сам блокирует push, если видит в нём похожий на ключ текст (это называется push protection). Но надеяться на это — всё равно что не пристёгиваться, потому что в машине есть подушка безопасности. Для параноиков есть `gitleaks` и `trufflehog` — они проверяют коммиты у вас на компьютере ещё до отправки.

# 2. Операторы и ловушки

Три слова, которые вы будете слышать постоянно, и которые многие путают всю жизнь.

**Выражение** — кусок кода, который *вычисляется* и даёт значение. `2 + 2` — выражение, его значение `4`. `temperature > 20` — выражение, его значение `True` или `False`.

**Оператор** — значок, который что-то делает со значениями. В `2 + 2` двойки — это *операнды*, а плюс — *оператор*. (В старых русских учебниках «оператором» называют ещё и `if`, и `while` — целые команды. Не путайтесь: у нас оператор — это то, что из значений делает новое значение. Чаще всего значок, иногда короткое слово: `and`, `or`, `not`, `in`, `is`. А `if` и `while` — это инструкции.)

**Инструкция** — целая команда, «предложение» на Python. `total = 250 * 2` — инструкция: она вычисляет выражение `250 * 2` и кладёт результат в переменную.

Из операндов и операторов мы лепим выражения, из выражений — инструкции, из инструкций — программы. Как из букв — слова, из слов — предложения, а из предложений — объяснительную в деканат.

## Арифметика, которую вы знаете, и которую не знаете

Плюс, минус, умножить и разделить вы видели в Л1. Напомню только одно: обычное деление `/` **всегда** возвращает дробное число, даже `10 / 2` даст `5.0`.

Целочисленное деление `//`, остаток `%` и степень `**` в Л1 тоже мелькали — одной строкой, между делом. Теперь разберёмся по-настоящему. На бананах. Смотрим.

````python
# --- Знакомое: обычное деление всегда даёт float ---
print(f"Обычное деление (результат всегда float): 10 / 3 = {10 / 3}")
print("-" * 20)

# --- Главные герои: //, %, ** ---

# 1. Целочисленное деление (Floor Division) //
# Показывает, сколько ЦЕЛЫХ раз одно число помещается в другое. Дробная часть отбрасывается.
bananas = 25
people = 7
print(f"У нас {bananas} бананов и {people} человек.")
print(f"Каждый получит по {bananas // people} банана.")

# 2. Остаток от деления (Modulo) %
# Показывает, что останется после целочисленного деления.
print(f"Остаётся {bananas % people} банана — 'налог для организатора' :)")

# Практическое применение %: проверка на четность
# Если число % 2 == 0, оно четное. Иначе - нечетное. Это используется ПОСТОЯННО.
number_to_check = 14
# if/else — «если… иначе…». Подробно разберём в разделе 6, пока просто читайте как по-русски
if number_to_check % 2 == 0:
    print(f"Число {number_to_check} - четное.")
else:
    print(f"Число {number_to_check} - нечетное.")

# 3. Возведение в степень (Exponentiation) **
print(f"\n2 в 10 степени = {2**10}")
print(f"Квадратный корень из 9 (это 9 в степени 0.5) = {9**0.5}")
````

Двадцать пять бананов на семерых: каждому по три, четыре остаётся — это налог для организатора. `//` отвечает на вопрос «сколько целых раз помещается», `%` — «сколько осталось».

Запомните проверку на чётность `x % 2 == 0`. Вы будете писать её чаще, чем своё имя. А `%` вообще — главный оператор для всего, что ходит по кругу: дни недели, часы на циферблате, «каждый третий покупатель получает скидку».

Вот вся арифметика одной табличкой:

| Оператор | Что делает | Пример | Результат |
|:---:|---|---|---|
| `+` | сложение | `4 + 5` | `9` |
| `-` | вычитание | `8 - 3` | `5` |
| `*` | умножение | `5 * 5` | `25` |
| `/` | деление — **всегда** float | `4 / 2` | `2.0` |
| `//` | целочисленное деление | `15 // 4` | `3` |
| `%` | остаток от деления | `7 % 2` | `1` |
| `**` | возведение в степень | `2 ** 3` | `8` |

### Деление с минусом: Python не такой, как C

А теперь вопрос залу. Сколько будет `-7 // 2`? Минус три с половиной, отбросим дробную часть — минус три, да?

````python
print(-7 // 2)   # целочисленное деление
print(-7 % 2)    # остаток
print(int(-7 / 2))  # а вот int() просто отрезает дробную часть
````

Минус четыре. Потому что `//` в Python округляет **вниз**, к минус бесконечности, а не к нулю. Минус три с половиной вниз — это минус четыре. А остаток подстраивается так, чтобы сходилось: `-4 * 2 + 1 = -7`.

В C, C++ и Java будет минус три и остаток минус один. Если когда-нибудь будете переписывать код с другого языка и у вас «на отрицательных числах всё поехало» — вспомните эту ячейку. Хорошая новость: в Python остаток от деления на положительное число всегда неотрицательный, поэтому проверка `x % 2 == 1` работает и для `-17`. В C — нет.

А `int()` действительно просто отрезает дробную часть, в сторону нуля. Три разных способа «округлить» — три разных ответа.

### Деление на ноль

На ноль делить нельзя. Python с этим согласен и падает с `ZeroDivisionError`. Как от этого защищаться, разберём в разделе про ветвление — там мы специально его уроним.



*Поделили на ноль — забанены в калькуляторе навсегда. Python гуманнее: всего лишь `ZeroDivisionError`.*

## `round()` врёт? Нет, это вы не знаете, как он работает

Округлим полтора и два с половиной. Что ждём? Два и три, как в школе.

````python
print(round(0.5), round(1.5), round(2.5), round(3.5))
````

Ноль, два, два, четыре. Школа плачет.

Это не баг, это **банковское округление**: если число ровно посередине, Python округляет к **чётному**. Зачем? Представьте банк, который миллион раз в день округляет копейки. Если половинки всегда округлять вверх, ошибка копится в одну сторону, и за год набегает приличная сумма — либо банк теряет, либо клиенты. А если половинки уходят то вверх, то вниз, ошибки взаимно гасятся. Так по умолчанию округляет арифметика с плавающей точкой (стандарт IEEE 754) и, например, функция `Round` в VBA. А вот обычная `ОКРУГЛ` в Excel и многие базы данных округляют половинки «от нуля», по-школьному. Поэтому одна и та же сумма в Excel и в Python может разойтись на копейку — и бухгалтер придёт именно к вам.

А теперь второй эффект, который все путают с первым. Округлим до одного знака три числа:

````python
print(round(2.65, 1))
print(round(2.75, 1))
print(round(2.85, 1))
````

2.65 превратилось в 2.6, 2.75 — в 2.8. «Ну вот, банковское правило, к чётному!» — скажет вам половина интернета, глядя на первые две строчки. А теперь третья: 2.85 превратилось в 2.9. Девятка, на минуточку, нечётная. По правилу «к чётному» должно было выйти 2.8. Значит, в первой и третьей строчке дело вообще не в чётности.

Посмотрим, какое число на самом деле лежит в памяти. `from decimal import Decimal` означает «возьми из модуля `decimal` одну нужную вещь — `Decimal`». А `Decimal(число)` показывает дробь со всеми цифрами, которые компьютер обычно от вас прячет:

````python
from decimal import Decimal

print(Decimal(2.65))
print(Decimal(2.75))
print(Decimal(2.85))
````

Вспоминаем Л1, где `0.3 - 0.2` не равнялось `0.1`. Компьютер хранит дроби в двоичном виде, и 2.65 в двоичной системе записать точно невозможно — как одну треть в десятичной. В памяти лежит `2.6499999…`, а это меньше половины, значит, честно округляется вниз. С 2.85 та же беда, только в другую сторону: в памяти `2.8500000…0888`, чуть больше половины — округляется вверх, и плевать ему на чётность. А вот 2.75 = 2 + 1/2 + 1/4 записывается в двоичном виде точно, это настоящая половинка — и только тут работает правило «к чётному».

Итого, у `round` два разных эффекта:
1. **Настоящие половинки** (`0.5`, `2.5`, `2.75`) — округляются к чётному.
2. **«Половинки», которых нет** (`2.65`, `2.85`, `2.675`) — округляются туда, куда скажет двоичное представление. Иногда это совпадает с «к чётному», иногда нет — лотерея.

Если вам нужно школьное округление — например, для денег — используйте модуль `decimal` и **передавайте числа строкой**, чтобы они не успели испортиться:

````python
from decimal import Decimal, ROUND_HALF_UP

price = Decimal("2.675")
print(price.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))
````

## 🧨 Экстра: биржа, которая потеряла половину индекса

Январь 1982 года. Ванкуверская фондовая биржа запускает новый индекс, стартовое значение — ровно 1000. Индекс пересчитывается после каждой сделки, примерно три тысячи раз в день, и каждый раз результат обрезается до трёх знаков после запятой. Не округляется — **обрезается**, лишние цифры просто выкидываются.

Каждое обрезание отнимает в среднем полтысячной. Мелочь. Но три тысячи раз в день, двадцать два месяца подряд. К ноябрю 1983 года индекс стоял на **524**. Рынок всё это время рос, а индекс показывал, что биржа потеряла почти половину стоимости. Когда ошибку нашли и пересчитали правильно, индекс за одни выходные «подскочил» с 524 до **1098**.

Мораль: способ округления — это не мелочь. Ошибка меньше тысячной, умноженная на миллион операций, — это уже катастрофа.

И чтобы совсем мрачно. 25 февраля 1991 года, война в Персидском заливе, город Дахран. Американский зенитный комплекс Patriot проработал без перезагрузки больше ста часов. Время в нём считалось десятыми долями секунды, а число 1/10 хранилось в 24-битном регистре — и, как мы теперь знаем, точно записать 0.1 в двоичном виде невозможно. Крошечная ошибка на каждом тике за сто часов накопилась в треть секунды. Иракская ракета Scud летит почти 1,7 километра в секунду — за треть секунды это больше полукилометра. «Окно», в котором радар искал ракету, по расчёту американской Счётной палаты (GAO) съехало на 687 метров. Patriot её просто потерял. Ракета попала в казарму, погибли 28 человек. Исправленная программа приехала на следующий день.

Вот так безобидное `0.3 - 0.2 == 0.1` → `False` из Л1 выглядит в реальной жизни.

````python
# Цикл for и import разберём позже (for — в разделе 9). Сейчас смотрите только на результат.
# Смоделируем Ванкувер: 3000 пересчётов в день, ~22 месяца.
# Каждое изменение индекса маленькое и случайное, но в среднем рынок стоит на месте.
import math
import random

random.seed(1982)
honest = cut = 1000.0
for _ in range(3000 * 22 * 21):          # ~21 торговый день в месяце
    change = random.uniform(-0.00005, 0.00005)
    honest = round(honest * (1 + change), 3)           # округляем
    cut = math.floor(cut * (1 + change) * 1000) / 1000  # обрезаем
print(f"Округляли: {honest:.3f}")
print(f"Обрезали:  {cut:.3f}")
````

Один и тот же «рынок», одни и те же сделки. Разница — только в том, как выкидывали четвёртый знак. У нас вышло даже хуже, чем в Ванкувере, — модель грубая, но эффект ровно тот же: обрезание всегда отнимает и никогда не добавляет.

## Составное присваивание

Возвращаемся к будничному. Очень часто нужно взять переменную, что-то с ней сделать и положить результат обратно. Для этого есть сокращения: `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`. Смотрим на банковском счёте:

````python
# Базовое присваивание
balance = 1000
print(f"Начальный баланс: {balance}")

# Мы хотим пополнить баланс на 500
# Длинный способ:
balance = balance + 500
print(f"Баланс после пополнения (длинный способ): {balance}")

# Короткий и предпочтительный способ с оператором +=
balance += 500  # Это полностью эквивалентно balance = balance + 500
print(f"Баланс после пополнения (короткий способ): {balance}")

# Такие же сокращения есть для всех арифметических операторов:
balance -= 300  # Списание
print(f"Баланс после списания 300: {balance}")

balance *= 1.1 # Начисление процентов (увеличение на 10%)
print(f"Баланс после начисления 10% годовых: {balance:.2f}")

balance /= 2 # Разделили деньги с другом
print(f"Баланс после дележки: {balance:.2f}")
````

`balance += 500` — ровно то же самое, что `balance = balance + 500`, просто короче. В Python, кстати, **нет** `balance++`, как в C. Не ищите: его тут никогда и не было — специально, чтобы не было загадок вроде `i++ + ++i` из C. Одинокий `balance++` даст `SyntaxError`. Ирония: сама загадка `i++ + ++i` в Python запустится и молча вернёт `2 * i` — для Python это просто `i` плюс пачка унарных плюсов. Ничего не увеличится. Добро пожаловать.

## Переменная — это ярлык

В Python значения — числа, строки, списки — живут сами по себе, где-то в памяти. Это **объекты**. А переменная — это **ярлык**, бумажка с именем, которую приклеили к объекту. Коробки нет. Есть объект и есть верёвочка от имени к нему.

Пока вы работаете с числами и строками, разницы не видно. Но вот пример, который ломает «коробочную» модель. Квадратные скобки — это список, мы разберём его позже, сейчас достаточно знать, что `.append()` добавляет элемент в конец.

````python
my_cart = ["хлеб", "молоко"]
wife_cart = my_cart            # «скопировали» корзину?
wife_cart.append("коньяк")
print(my_cart)
````

Вы ничего не добавляли в свою корзину, а коньяк там есть. Потому что строчка `wife_cart = my_cart` ничего не скопировала. Она приклеила второй ярлык к **той же самой** корзине. Корзина одна, имени два.

Если бы переменная была коробкой, у нас было бы две коробки. А так — одна корзина и два человека, которые в неё кладут. Запомните эту картинку: она объяснит вам кучу странных багов в будущем. Как делать настоящие копии — расскажу в разделе про списки.

## Динамическая типизация

В C++ или Java вы обязаны заранее объявить: в этой переменной будут только целые числа.
```cpp
int user_age = 25;           // здесь ВСЕГДА будут целые числа
user_age = "двадцать пять";  // ошибка компиляции
```
В Python ярлык можно в любой момент переклеить на объект другого типа. Тип есть у **объекта**, а у ярлыка его нет:

````python
# Одна и та же переменная может хранить данные разных типов в разное время
my_variable = 100
print(f"Сейчас в переменной лежит {my_variable}, ее тип - {type(my_variable)}")

my_variable = "Сто рублей"
print(f"А теперь в ней лежит '{my_variable}', ее тип - {type(my_variable)}")

my_variable = True
print(f"А сейчас - {my_variable}, ее тип - {type(my_variable)}")
````

Гибко? Очень. Опасно? Тоже очень. Python не мешает вам положить в переменную `price` строку `"дорого"` — он просто упадёт потом, когда вы попытаетесь её умножить на количество.



*C — статическая и слабая. JavaScript — динамическая и слабая. Python — динамическая, но сильная: ярлык переклеить можно, а сложить строку с числом — нельзя.*

При этом не путайте «динамическая» и «слабая». Python **сильно** типизирован: он никогда не будет молча превращать строку в число, как это делает JavaScript. `"5" + 5` в JavaScript даст `"55"`, а Python скажет «так нельзя» и упадёт с `TypeError`. Мы на эту ошибку ещё полюбуемся в разделе 7. Сначала явно приведите тип, потом складывайте:

````python
print(int("5") + 5)
print("5" + str(5))
print("5" * 3)        # а вот умножение строки на число — законно: повторить 3 раза
````

## Как называть переменные

Компьютеру всё равно, как вы назовёте переменную. А вот человеку, который будет читать ваш код через полгода — то есть вам, — не всё равно. Для этого есть **PEP 8** — официальный свод правил хорошего тона в Python.

1. **`snake_case`**: маленькие буквы, слова через подчёркивание. `user_name`, `order_total`, `is_active`. А `userName` — это стиль из JavaScript и Java, у нас так не пишут. `UserName` с большой буквы — для классов, до них доберёмся в Л4.
2. **Имя объясняет содержимое.** `user_age = 30` — понятно. `x = 30` — что за `x`? Исключение — счётчики в коротких циклах (`i`, `j`) и математика (`x`, `y`).
3. **Английский.** Можно писать имена хоть по-русски, Python это разрешает: `возраст = 20` — рабочий код. Но если вы так будете делать, вы получите ачивку **«У меня нет друзей»**. И транслит `vozrast` — не лучше.
4. **Не называйте переменные `l` и `O`** — в половине шрифтов их не отличить от `1` и `0`.

Технически имя может состоять из букв, цифр и подчёркиваний, но не может начинаться с цифры и не может совпадать с **ключевым словом** языка. Ключевых слов немного, Python может сам их показать:

# 4. Строки

Строки вы уже видели: всё, что в кавычках. Но большая часть реальной работы аналитика — это не формулы, а возня с текстом: имена клиентов с лишними пробелами, города в трёх регистрах, даты в формате «как бог на душу положит». Поэтому строки разберём основательно, одним большим блоком.

## Строка — это улица с домами

Точное определение: строка — это **упорядоченная последовательность символов**. Ключевое слово — «упорядоченная». У каждого символа есть свой номер, и порядок их не меняется.

Представьте улицу. Каждый символ — дом, у каждого дома есть номер. Знаете номер — можете постучаться в любой дом. Только одна особенность: на этой улице нумерация домов начинается **с нуля**. Почему? Исторически номер — это не «какой по счёту», а «на сколько шагов отойти от начала улицы». От первого дома до первого дома — ноль шагов.

## Кавычки

Одинарные и двойные кавычки в Python — одно и то же. Выбирайте те, которых нет внутри текста. Тройные кавычки — для многострочного текста:

````python
# Одинарные кавычки
string_1 = 'Привет, мир!'
print(string_1)

# Двойные кавычки
string_2 = "He said, 'Python is awesome!'"
print(string_2)

# Тройные кавычки для многострочного текста
multi_line_string = """Это первая строка.
А это вторая.
    И даже отступы сохраняются!
"""
print(multi_line_string)

# Тройные кавычки для текста с разными кавычками внутри
complex_string = '''"It's a complex case", she said.'''
print(complex_string)
````

Кстати, строку в тройных кавычках, не присвоенную никакой переменной, иногда используют как многострочный комментарий. Формально это не комментарий, а строка, которую Python создал и тут же выбросил. Настоящий комментарий в Python один — решётка `#` до конца строки.

## Индексы: постучаться в один дом

Вот улица `"Python"` с номерами домов. Нижняя строчка — номера с конца улицы, о них чуть ниже:

| Символ | `P` | `y` | `t` | `h` | `o` | `n` |
|---|:---:|:---:|:---:|:---:|:---:|:---:|
| Индекс | `0` | `1` | `2` | `3` | `4` | `5` |
| С конца | `-6` | `-5` | `-4` | `-3` | `-2` | `-1` |

Номер дома пишем в квадратных скобках сразу после строки. Стучимся:

````python
word = "Python"

# Получаем первый символ (индекс 0)
first_char = word[0]
print(f"Первый символ слова '{word}': {first_char}")

# Получаем третий символ (индекс 2)
third_char = word[2]
print(f"Третий символ слова '{word}': {third_char}")

# Получаем последний символ (индекс -1)
last_char = word[-1]
print(f"Последний символ слова '{word}': {last_char}")

# А word[10] упадёт с IndexError — дома №10 на этой улице нет. Уроним в разделе 7.
````

Отрицательные индексы считают с конца: `-1` — последний символ, `-2` — предпоследний. Очень удобно: не надо знать длину строки, чтобы взять последнюю букву.

## Срезы: вырезать целый квартал

Срез — это не один дом, а кусок улицы: `строка[start:stop:step]`.

- `start` — откуда начинаем, **включительно**. Не указали — с начала.
- `stop` — где заканчиваем, **не включительно**. Не указали — до конца.
- `step` — шаг. Не указали — 1.

Почему `stop` не включается? Чтобы `phrase[:6]` и `phrase[6:]` вместе давали ровно всю строку, без дыр и без повторов. И чтобы длина среза `[2:5]` считалась в уме: 5 − 2 = 3 символа. Это неудобно ровно первую неделю, потом вы будете благодарны.

````python
phrase = "Анализ данных"
# Индексы: 0123456789...

# Срез с 0 по 5 символ (6-й не включается)
slice_1 = phrase[0:6]
print(f"Срез [0:6]: '{slice_1}'")

# Если начинаем с 0, `start` можно опустить
slice_2 = phrase[:6]
print(f"Срез [:6]: '{slice_2}'")

# Срез с 7 символа и до конца
slice_3 = phrase[7:]
print(f"Срез [7:]: '{slice_3}'")

# Срез с шагом 2 (берем каждый второй символ)
every_second_char = phrase[::2]
print(f"Каждый второй символ: '{every_second_char}'")

# ❗️Самый известный трюк со срезами: переворот строки
reversed_phrase = phrase[::-1]
print(f"Перевернутая строка: '{reversed_phrase}'")
````

`[::-1]` — шаг минус один, идём с конца в начало. Самый известный трюк со срезами, его спрашивают на каждом втором собеседовании для джунов.

И одна важная разница между индексом и срезом. Индекс за пределами строки — это ошибка. А срез за пределами — просто пустая строка или «сколько есть». Срез прощает, индекс — нет:

````python
dna = "ATTCGGAGCT"
print(dna[1:4], dna[4:], dna[1:-1], dna[1:-1:2])
print(repr(dna[100:]))   # за пределами строки — пустая строка, без ошибки
print(dna[:100])         # просим 100 символов — дают сколько есть
````

### Палиндром

Палиндром — фраза, которая одинаково читается с обеих сторон. Проверить это срезом — одна строчка: сравнить строку с ней же, перевёрнутой. Классика — «А роза упала на лапу Азора». Введите её:

`False`. Ломается сразу на первой букве: «А» большая, а «а» в конце — маленькая. И пробелы стоят не на тех местах. Компьютер не знает, что такое «читается одинаково», он сравнивает символы. Значит, сначала нужно привести строку к нормальному виду: убрать пробелы и сделать все буквы маленькими.

````python
clean = phrase.lower().replace(" ", "")
print(clean)
print(clean == clean[::-1])
````

Вот теперь `True`. Запомните этот порядок действий, он универсален: **сначала чистим данные, потом сравниваем.** Уже на следующей лекции вы будете делать то же самое с файлами логов, а во втором семестре — с миллионами строк.

## Методы: строка умеет кое-что сама

В Python почти всё — объекты. А у объекта есть встроенные умения, которые вызываются через точку: `строка.умение()`. Эти умения называются **методами**. Один вы уже видели в Л1 — `.upper()`. Строка умеет сама сделать из себя заглавные буквы, убрать пробелы по краям, заменить одно слово другим, посчитать вхождения.

````python
s = "   Привет, МИСИС! Привет, Python!   "

print(repr(s.strip()))                    # убрать пробелы по краям
print(s.upper())                          # всё заглавными
print(s.lower())                          # всё маленькими
print(s.strip().replace("Привет", "Пока"))  # заменить
print(s.count("Привет"))                  # сколько раз встречается
print("иван иванович иванов".title())      # каждое слово с большой буквы
print("report_2026.csv".startswith("report_"), "report_2026.csv".endswith(".csv"))
print("2026".isdigit(), "1200р".isdigit(), "3450.50".isdigit())  # только цифры? точка — не цифра
````

`repr()` в первой строке показывает строку вместе с кавычками — так видно, есть ли по краям пробелы. Полезная штука для отладки: пробел в конце строки на глаз не видно, а сравнение из-за него ломается.

## Строки неизменяемы

А теперь очень важная вещь, на которой спотыкаются все. Методы **не меняют** исходную строку. Они возвращают **новую**.

Строка — как фотография. Вы можете её отфотошопить и сохранить как новый файл, но сам исходный снимок останется прежним. Это свойство называется **иммутабельность** (неизменяемость).

````python
original_phrase = "Я люблю спать и кушать"
print(f"Исходная строка (до методов): {original_phrase}")

# Вызываем метод, но не сохраняем результат в переменную
original_phrase.upper()

print(f"Исходная строка (после вызова .upper()): {original_phrase}") # Ничего не изменилось!

# Чтобы изменение "сохранилось", нужно перезаписать переменную
original_phrase = original_phrase.upper()
print(f"Исходная строка (после перезаписи): {original_phrase}")
````

Во второй раз ничего не изменилось. `.upper()` честно создал строку капсом — и она тут же улетела в никуда, потому что мы её никуда не сохранили. Работает только `original_phrase = original_phrase.upper()`: создать новую строку и переклеить на неё ярлык.

Ещё одно следствие — нельзя поменять одну букву:

````python
word = "кот"
# word[0] = "р"    # TypeError: 'str' object does not support item assignment
word = "р" + word[1:]
print(word)
````

## Очистка данных: то, ради чего всё это

Вот пример из реальной жизни. Пользователь ввёл город. С пробелами по краям, с переносом строки в конце — потому что скопировал из Excel. Приводим его в порядок. А дальше два главных метода аналитика: `.split()` режет строку на список кусков, `.join()` склеивает список обратно.

````python
# --- Очистка и форматирование ---
messy_data = "   Москва, Россия   \n"
print(f"Исходные данные: '{messy_data}'")

# 1. Убираем лишние пробелы и переносы строк
clean_data = messy_data.strip()
print(f"После .strip(): '{clean_data}'")

# 2. Приводим к единому регистру
lower_case_data = clean_data.lower()
print(f"После .lower(): '{lower_case_data}'")

# 3. Заменяем запятую на дефис
replaced_data = lower_case_data.replace(",", " -")
print(f"После .replace(): '{replaced_data}'")
print("-" * 30)

# --- Трансформация: split и join ---

# Задача: из строки "яблоко,банан,вишня" получить список ['яблоко', 'банан', 'вишня']
csv_string = "яблоко,банан,вишня"
fruits_list = csv_string.split(',') # Разделяем по запятой
print(f"Результат .split(','): {fruits_list}")
print(f"Первый элемент списка: {fruits_list[0]}")

# Обратная задача: из списка `fruits_list` собрать строку, разделенную пробелами
fruits_to_join = ['яблоко', 'банан', 'вишня']
# ' ' - это строка-разделитель. Мы вызываем у нее метод .join()
joined_string = " ".join(fruits_to_join)
print(f"Результат ' '.join(...): '{joined_string}'")
````

Обратите внимание на странный синтаксис `" ".join(...)`: метод вызывается у **разделителя**, а не у списка. Все сначала пишут `fruits.join(" ")`, получают ошибку и запоминают навсегда.

## f-строки: итоговая версия

В Л1 вы видели f-строки: буква `f` перед кавычками, переменные в фигурных скобках. Теперь — их суперсилы. Внутри фигурных скобок можно писать любое выражение, а после двоеточия — формат. `:.2f` — два знака после запятой, `:>10` — выровнять по правому краю в поле ширины 10. Чек за арбуз:

````python
product_name = "арбуз"
price = 120.5555
weight = 8.765

receipt = f"""
==============================
Товар: {product_name.upper()}
Цена: {price:.2f} руб/кг
Вес: {weight:.1f} кг
------------------------------
Итого: {(price * weight):.2f} руб.
==============================
"""
print(receipt)
````



*Пуш от маркетплейса: «${item_name} за ${item_price}». Значения забыли подставить. Python-версия того же позора — забыть букву `f`: без неё `"{price}"` — это просто фигурные скобки.*

И ещё одна суперсила, ради которой стоит полюбить f-строки. Знак равенства после выражения печатает и само выражение, и его значение. Это лучший друг при отладке:

````python
price = 120.5555
weight = 8.765
print(f"{price=}")
print(f"{price * weight = :.2f}")
print(f"{len('арбуз')=}")
````

Вместо `print("price =", price)` — просто `print(f"{price=}")`. Каждый раз, когда вы не понимаете, что лежит в переменной, — вот этот приём.

> 🤖 **Где ошибается ваш ИИ.** Многие модели до сих пор уверены, что внутри f-строки нельзя использовать те же кавычки, что и снаружи, и будут переписывать ваш `f"{user["name"]}"` на `f"{user['name']}"`. Это правило умерло в Python 3.12 (2023 год). Начиная с 3.12 так писать можно. Модели учились на старом коде, а кода на 3.12+ в интернете пока меньше.

````python
user = {"name": "Аня"}   # это словарь, о нём — в разделе 12
print(f"Привет, {user["name"]}!")   # в Python 3.12+ — законно
````

## Обратный слэш, который всё ломает

Некоторые символы в строку просто так не напишешь: перенос строки, табуляцию, кавычку внутри кавычек. Для этого есть **экранирование** — обратный слэш и буква после него:

````python
print("Первая строка\nВторая строка")   # \n — перенос строки
print("Имя:\tАня")                     # \t — табуляция
print("Он сказал: \"Привет\"")          # \" — кавычка внутри строки
print("Обратный слэш: \\")              # \\ — сам обратный слэш
print("Длина '\\n':", len("\n"))         # \n — это ОДИН символ
````

А теперь грабли, на которые наступает каждый первый пользователь Windows. Вы хотите прочитать файл и пишете путь к нему:

````python
path = "C:\data\new"
print(path)
````

Первое: Python выдал **SyntaxWarning: invalid escape sequence '\d'** (Jupyter может показать его два-три раза — это одно и то же предупреждение). Комбинации `\d` не существует, и Python предупреждает: «Вы точно знаете, что делаете?» Начиная с Python 3.12 это предупреждение, а в будущих версиях станет ошибкой. На Python 3.14 и новее текст длиннее и заботливее: `"\d" is an invalid escape sequence. Such sequences will not work in the future. Did you mean "\\d"? A raw string is also an option.` — Python сам подсказывает лечение, до которого мы сейчас дойдём.

Второе, и это хуже: `\n` — это существующая комбинация. Перенос строки. Поэтому путь `C:\data\new` превратился в `C:\data` и `ew` на новой строке. Файл не найдётся, и вы будете полчаса смотреть на совершенно правильный путь.

Лечение — **сырая строка**: буква `r` перед кавычками. Внутри неё обратный слэш — это просто обратный слэш, ничего не экранируется:

````python
path = r"C:\data\new"
print(path)

path = "C:/data/new"   # или прямые слэши — Windows их тоже понимает
print(path)
````

А вообще с путями к файлам по-взрослому работают через модуль `pathlib` — расскажу в следующей лекции.

> 🤖 **Где ошибается ваш ИИ.** Попросите ассистента отсортировать список русских слов. Он напишет `sorted(words)` и скажет «по алфавиту». А потом в вашем списке появится «ёлка» — и окажется после «яблока». Python сравнивает строки не по алфавиту, а по **номерам символов в Unicode**. Буква «ё» стоит в таблице после «я», а «Ё» — перед «А». И заглавные буквы все идут раньше строчных.

````python
print(sorted(["яблоко", "ёлка", "еда", "Ёж", "Арбуз", "арбуз"]))   # sorted() — отсортировать, подробно в разделе 9
print(ord("е"), ord("ё"), ord("я"))   # ord() — номер символа в таблице Unicode
print("ёлка" > "яблоко")
````

Номер «ё» — 1105, номер «я» — 1103. Поэтому «ёлка» больше «яблока». Для настоящей сортировки по-русски есть модуль `locale` или хитрый ключ сортировки — к ключам сортировки мы придём в следующей лекции. А пока просто помните: «по алфавиту» и «по возрастанию строк» — это не одно и то же.



*Подменить другу в коде `;` на греческий знак вопроса `;` — выглядят одинаково, а номера в Unicode разные: 59 (U+003B) против 894 (U+037E). Даже дьявол считает, что это перебор.*

## 📚 Самостоятельно: форматирование через `%` и `.format()`

f-строки появились в 2016 году. До них строки форматировали двумя способами, и вы их встретите в старом коде, в документации и в ответах на StackOverflow:

````python
name = "МИСИС"
score = 42.7250
print("Привет, %s! Балл: %.2f" % (name, score))        # самый старый, «как в C»
print("Привет, {}! Балл: {:.2f}".format(name, score))  # .format()
print(f"Привет, {name}! Балл: {score:.2f}")            # f-строка — используем её
````

Правила форматирования после двоеточия у `.format()` и f-строк одинаковые, а у `%` — свои. Всё это описано в документации: [Format Specification Mini-Language](https://docs.python.org/3/library/string.html#formatspec). Писать новый код через `%` не нужно, но читать его — да. Одно место, где `%` живёт до сих пор, — модуль `logging`, вы с ним встретитесь при написании ботов.

# 5. Логика: как научить программу думать

До сих пор наши программы были поварами, которые тупо следуют рецепту: возьми, положи, перемешай. Никаких решений. А настоящей программе постоянно надо выбирать: «если пользователь совершеннолетний — пустить, иначе — нет», «если температура ниже нуля **и** идёт снег — включить отопление», «если ввели "выход" **или** нажали Esc — закрыться».

Для этого нужны две вещи: **сравнения** — чтобы задавать вопросы о данных, — и **логические операторы**, чтобы склеивать вопросы в сложные условия.

## Сравнения

Шесть операторов, и все возвращают `True` или `False`:

````python
print(5 == 5)    # равно
print(5 != 6)    # не равно
print(5 > 3, 5 < 3)
print(5 >= 5, 5 <= 4)
````

Главная ошибка первой недели: **`=` и `==` — это разные вещи**. Один знак равенства — «приклей ярлык», два — «спроси, равны ли». Если в `if` вы напишете один знак, Python, к счастью, упадёт с SyntaxError. В C, кстати, не упадёт, и там из-за этого десятилетиями горели целые системы.

Строки тоже сравниваются: посимвольно, по номерам в Unicode — как мы только что видели на «ёлке». Регистр важен.

````python
print("яблоко" == "яблоко")
print("Python" == "python")
print("abc" > "ab")      # длиннее при равном начале — больше
print("abc" > "db")      # сравнение решает первый же различающийся символ
````

## Цепочки сравнений

Проверить, что число лежит в промежутке. В большинстве языков пишут `x >= 3 and x < 6`. А в Python можно как в математике:

````python
x = 4
print(3 <= x < 6)
print(1 < 2 < 3 < 4 < 5)
print(10 < x < 20)
````

`3 <= x < 6` — это ровно `3 <= x and x < 6`, только читается как в учебнике алгебры. Пользуйтесь.

## `and`, `or`, `not`

- `A and B` — истинно, только если истинны **оба**. Чтобы получить права, нужно `(возраст >= 18) and (вы хотя бы немного адекватны — есть медсправка)`.
- `A or B` — истинно, если истинно **хотя бы одно**. Скидка, если вы `студент or у вас карта лояльности`.
- `not A` — переворачивает: `not True` — это `False`.

````python
age = 19
has_license = True
print("Можно арендовать машину?", age >= 21 and has_license)

is_student = False
has_card = True
print("Будет скидка?", is_student or has_card)

is_weekend = True
print("Надо на работу?", not is_weekend)
````

Порядок: сначала `not`, потом `and`, потом `or` — как умножение раньше сложения. Сомневаетесь — ставьте скобки. Скобки бесплатные, а баг из-за приоритета будете искать час.

## Истинность: что Python считает ложью

А теперь то, о чём молчат учебники для начинающих. В условии можно писать не только `True` и `False`, а **что угодно**. Python сам решит, «истинно» оно или нет. Правило простое: **ложью считается всё пустое и всё нулевое.** Остальное — истина.

````python
print(bool(0), bool(0.0), bool(""), bool([]), bool({}), bool(None))
print(bool(42), bool(-1), bool("0"), bool(" "), bool([0]), bool("False"))
````

Первая строка — сплошные `False`: ноль, пустая строка, пустой список, пустой словарь, `None`. Вторая — сплошные `True`, включая строку `"0"` (она не пустая — в ней символ ноль), строку из одного пробела и, моё любимое, строку `"False"`. Строка `"False"` — это истина. Добро пожаловать в программирование.

Зачем это нужно? Чтобы писать коротко и по-человечески. Вместо `if len(items) == 0:` пишут:

````python
items = []
if not items:
    print("Корзина пуста")
````

«Если нет товаров» — читается как обычный русский язык.

## `and` и `or` возвращают не `True`, а операнд

Вот вам факт, который знает, дай бог, треть программистов на Python. Многие учебники пишут, что логические операторы «всегда возвращают `True` или `False`». Для `not` — да. А для `and` и `or` — **нет**. Они возвращают **один из операндов**.

- `A or B`: если `A` истинно — возвращается `A`, иначе — `B`.
- `A and B`: если `A` ложно — возвращается `A`, иначе — `B`.

````python
print(5 and 7)
print(0 and 7)
print("" or "Аноним")
print("Аня" or "Аноним")
print([] or "пусто")
````

И это не теоретическая муть, а самый частый приём для значения по умолчанию. Запустите ячейку и просто нажмите Enter, ничего не вводя:

````python
name = input("Как вас зовут? ") or "Аноним"
print(f"Привет, {name}!")
````

Пустая строка — ложь, поэтому `or` вернул правую часть. Одна строчка вместо четырёх с `if`.

## Ленивые вычисления

И последнее. Python ленивый, и это прекрасно. В `A and B`, если `A` уже ложно, Python **даже не смотрит** на `B` — результат и так ложь. В `A or B`, если `A` истинно, `B` тоже не вычисляется. Это называется короткое замыкание. Смотрите, как оно спасает от деления на ноль:

````python
divisor = 0
result = (divisor != 0) and (10 / divisor > 1)
print(result)
````

Если бы Python вычислил правую часть, мы бы упали с `ZeroDivisionError`. Но левая часть оказалась ложной, и до деления дело не дошло. Порядок условий в `and` — это не косметика: сначала проверка «а можно ли», потом само действие.

## 📚 Самостоятельно: таблицы истинности

Для тех, кто любит строгость. Все варианты для `and` и `or` — их удобно не зубрить, а сгенерировать:

````python
print(f"{'A':<6} {'B':<6} {'A and B':<8} {'A or B':<7}")
for a in (True, False):
    for b in (True, False):
        print(f"{a!s:<6} {b!s:<6} {(a and b)!s:<8} {(a or b)!s:<7}")
````

Циклы `for` мы разберём через пару разделов, а `!s` внутри f-строки значит «преобразовать в строку перед форматированием». Сверьте таблицу с правилами выше: `and` даёт `True` в одной строке из четырёх, `or` — `False` в одной строке из четырёх. И ещё два закона, которые полезно знать: `not (A and B)` — то же самое, что `not A or not B`, а `not (A or B)` — то же, что `not A and not B`. Это законы де Моргана, они пригодятся, когда условие в `if` станет нечитаемым и его захочется вывернуть наизнанку.

# 6. Ветвление: развилка на дороге

Ваша программа — путник на дороге. До сих пор дорога была прямой. Теперь начинаются развилки с указателями: «если пойдёшь налево — коня потеряешь, направо — сам пропадёшь». Указатели в Python называются `if`, `elif` и `else`.

## Священное правило: отступы

Прежде чем писать первый `if`, главное правило синтаксиса Python, которое отличает его почти от всех языков.

В C++, Java, JavaScript блок кода внутри условия выделяют фигурными скобками:
```javascript
if (age >= 18) {
  console.log("Доступ разрешён");
}
```
В Python фигурных скобок нет. Вместо них — **отступ**. Код, который относится к условию, сдвигается вправо. Всё, что сдвинуто одинаково, — один блок. Закончился сдвиг — закончился блок.

Это не рекомендация для красоты. Сам отступ — жёсткое правило синтаксиса. Ошибётесь с ним — получите `IndentationError`, или, что хуже, программа запустится, но будет делать не то.

А сколько пробелов? Синтаксису, строго говоря, всё равно: два, четыре, семь — лишь бы внутри одного блока одинаково. Но есть PEP 8, общий стандарт оформления, и он говорит: **4 пробела**. Так пишет весь мир. Код с двумя пробелами заработает, но коллеги будут смотреть на вас как на человека в носках с сандалиями. Хорошая новость: VS Code сам ставит 4 пробела после двоеточия и по клавише `Tab`. Плохая: если вы скопировали код из мессенджера, где отступы — это табуляция вперемешку с пробелами, будет весело.

## `if`: сделать, только если

Перед вами самая защищённая система в мире. Пароль `666666`, введите его:

````python
user_password = "666666"
entered_password = input("Введите пароль для входа: ")

# Проверяем условие: введенный пароль совпадает с сохраненным?
if entered_password == user_password:
    # Этот блок выполнится ТОЛЬКО при совпадении паролей
    print("Аутентификация успешна!")
    print("Добро пожаловать в систему.")

print("\nПрограмма завершила работу.") # Эта строка выполнится всегда
````

Двоеточие в конце строки с `if` — обязательно. Две строки с отступом выполнились, потому что пароль совпал. Последняя строка без отступа выполнится всегда, она уже не относится к `if`.

## `if` + `else`: или-или

Два взаимоисключающих варианта: если условие истинно — делаем это, иначе — то. Введите возраст, например 17:

````python
user_age = int(input("Введите ваш полный возраст: "))

if user_age >= 18:
    print("✅ Доступ к контенту разрешен.")
    print("Вы можете продолжить.")
else:
    years_to_wait = 18 - user_age
    print("❌ Доступ запрещен. Вам еще нет 18.")
    print(f"Пожалуйста, возвращайтесь, когда подрастёте. Ждать осталось (лет): {years_to_wait}.")
````

## Деление на ноль: от наивного кода к нормальному

А теперь живой пример того, как пишется программа на самом деле. Напишем деление двух чисел — проще некуда. Введите 10, потом 0.

````python
a = int(input("Делимое: "))
b = int(input("Делитель: "))
print(a / b)   # ⚠️ упадёт специально, если ввести 0
````

`ZeroDivisionError: division by zero`. Программа умерла. Пользователь, который случайно ввёл ноль, видит красную простыню. Так программы писать нельзя.

Добавляем проверку: если делитель не ноль — делим, иначе — вежливо отказываем. Числа возьмём те же, что вы только что ввели:

````python
if b != 0:
    print(a / b)
else:
    print("Деление невозможно")
````

Уже лучше. Но можно ещё вежливее: дать человеку второй шанс. А если он и во второй раз введёт ноль — значит, сам виноват. Введите опять 0, не стесняйтесь:

````python
if b != 0:
    print(a / b)
else:
    print("Деление невозможно")
    b = int(input("Введите ненулевой делитель: "))
    if b == 0:
        print("Вы не справились!")
    else:
        print(a / b)

print("Продолжаем работу")
````

Видите `if` внутри `else`? Это вложенное условие, у него отступ уже 8 пробелов. Вкладывать можно сколько угодно, но больше двух-трёх уровней читать невозможно. Если у вас получилась лесенка, уходящая за правый край экрана — это знак, что код пора переписать. Как именно — сейчас покажу.

А что будет, если вместо числа ввести слово «десять»? `int("десять")` упадёт с `ValueError`. Это тоже надо ловить, и мы научимся — в следующей лекции, через `try`/`except`.

## `elif`: цепочка проверок

Что, если вариантов больше двух? Для этого есть `elif` — сокращение от «else if», «иначе, если».

Python проверяет условия **сверху вниз** и выполняет **первое** истинное. Остальные даже не смотрит. Если ни одно не сработало — выполняется `else`, если он есть. В любой цепочке `if-elif-else` выполняется **не больше одного** блока. Проверим на погоде. Введите `-80`:

````python
temperature = int(input("Введите температуру на улице (в градусах Цельсия): "))

if temperature > 25:
    print("Очень жарко! Наденьте шорты и кэпку.")
elif temperature > 15: # Эта проверка сработает, только если temperature <= 25
    print("Тепло. Джинсы и лёгкая куртка подойдут.")
elif temperature > 0:  # Эта проверка сработает, только если temperature <= 15
    print("Прохладно. Надевайте тёплое пальто и шапку.")
elif temperature > -25:
    print("Морозно! Нужен пуховик, шарф и перчатки.")
elif temperature > -75:
    print("Холодно! Шуба и валенки")
else: # Этот блок сработает, если temperature <= -75
    print("Экстремальный холод! Водка замерзла. Грустно. Лучше остаться дома.")
````

Обратите внимание на комментарии: проверка `temperature > 15` сработает, только если предыдущая `> 25` была ложной. Поэтому писать `15 < temperature <= 25` не нужно — верхнюю границу уже отсёк предыдущий `elif`.

### Почему не просто несколько `if`

Вот задача. По возрасту пользователя рекомендуем ему фильм. Введите `7`:

````python
age = int(input('Ваш возраст: '))

print('Рекомендовано:', end=' ')

if 3 <= age < 6:
    print('"Заводной апельсин"')
if 6 <= age < 12:
    print('"Зелёный слоник"')
if 12 <= age < 16:
    print('"Фиксики"')
if 16 <= age:
    print('"Гравити фолз"')
````

`end=' '` в `print` значит «в конце не переводить строку, а поставить пробел» — поэтому фильм допечатался в ту же строку, что и «Рекомендовано:».

Работает. Но плохо. Здесь четыре **независимых** `if`, и Python проверит все четыре, даже если первое уже сработало. Для четырёх условий не страшно. А теперь представьте, что каждая проверка — это запрос в базу данных на две секунды.

Можно вложить `if` в `else`, `if` в `else`, и так далее — получится та самая лесенка, уходящая вправо:
```python
if 3 <= age < 6:
    ...
else:
    if 6 <= age < 12:
        ...
    else:
        if 12 <= age < 16:
            ...
```
В C и Java эту лесенку записывают как `else if` в одну строчку. А в Python для неё есть отдельное слово — `elif`. Та же логика, но плоско:

````python
# возраст берём из прошлой ячейки

print('Рекомендовано:', end=' ')

if 3 <= age < 6:
    print('"Прощай, унылый учитель"')
elif 6 <= age < 12:
    print('"Помпоко: Война тануки в период Хэйсэй"')
elif 12 <= age < 16:
    print('"Ковбой Бибоп"')
elif 16 <= age:
    print('"Триган"')
````

Каждое следующее условие проверяется, только если все предыдущие ложны. Как только что-то сработало — выходим из всей конструкции.

Вот схема того, как поток выполнения проходит через `if-elif-else`:



*`else if` из C и Java против `elif`. Четыре буквы вместо шести — и лесенка превращается в ровный список.*

### Ловушка порядка

А теперь ошибка, которую делают все. Порядок условий в `elif` имеет значение. Студент получил 95 баллов:

````python
# Неправильный порядок проверки оценок
grade = 95

if grade > 50: # 95 > 50 -> True. Это условие сработает первым!
    print("Удовлетворительно.")
elif grade > 70:
    print("Хорошо.")
elif grade > 90:
    print("Отлично!")

# Программа выведет "Удовлетворительно.", хотя оценка 95.
# Правильно всегда идти от более строгих проверок к более общим.
````

«Удовлетворительно» за 95 баллов. Студент идёт в деканат, а вы — искать работу. Первым сработало самое общее условие `grade > 50`, и до остальных Python просто не дошёл.

Правило: **в цепочке `elif` идите от самого строгого условия к самому общему.** Сначала `> 90`, потом `> 70`, потом `> 50`.

## Тернарный оператор: `if` в одну строку

Очень часто `if-else` нужен, только чтобы положить в переменную одно из двух значений:

````python
# Обычный способ
user_age = 22
if user_age >= 18:
    status = "совершеннолетний"
else:
    status = "несовершеннолетний"
print(f"Статус пользователя (обычный способ): {status}")
````

Для этого случая есть короткая запись: `значение_если_да if условие else значение_если_нет`. Читается почти как английский: «совершеннолетний, если возраст ≥ 18, иначе — несовершеннолетний».

````python
status = "совершеннолетний" if user_age >= 18 else "несовершеннолетний"
print(status)

balance = -500
color = "green" if balance >= 0 else "red"
print(color)

x = 1
print("Адын" if x == 1 else "Не Адын")
````

Только не увлекайтесь. Тернарный оператор внутри тернарного оператора — это уже не элегантность, а покушение на коллег.

## `match`/`case`: когда вариантов много

С Python 3.10 есть ещё одна конструкция — `match`. Она удобна, когда вы сравниваете **одно значение** с кучей вариантов. Например, меню бота: пользователь присылает команду, мы отвечаем.

````python
command = "кофе"

match command:
    case "кофе":
        print("Варим кофе. 200 ₽")
    case "чай" | "чай зелёный":
        print("Завариваем чай. 150 ₽")
    case "меню" | "помощь" | "?":
        print("Доступно: кофе, чай")
    case _:
        print(f"Не знаю, что такое «{command}»")
````

`match command` — «сопоставляем вот это». Каждый `case` — один вариант. Вертикальная черта `|` значит «или»: `case "чай" | "чай зелёный"` сработает на любое из двух. А `case _` — это «всё остальное», как `else`.

Поменяйте `command` на `"?"` или `"водка"` и перезапустите.

Ровно то же можно написать через `if-elif` с `==` и `or`. `match` просто читается приятнее, когда вариантов больше трёх. А ещё он умеет много хитрого — разбирать списки и словари на части — но это мы оставим на потом. Одно предупреждение: в `case` пишите **значения** — строки, числа. Если написать там голое имя переменной, `case` его не сравнит, а перезапишет — и сработает на всё подряд. Python это заметит, только если после такого `case` стоят ещё варианты: тогда он откажется запускать код с `SyntaxError: ... makes remaining patterns unreachable`. А в последнем `case` такая ошибка пройдёт молча.



*Элмо, впервые увидевший `match`/`case` после двадцати `elif` подряд. Осторожно, вызывает привыкание.*

## Как читать сообщение об ошибке

Красная простыня, которую выдаёт Python, называется **трейсбек** (traceback). Первое правило: **читайте его снизу вверх.**

- **Последняя строка** — самое важное: тип ошибки и человеческое описание. `NameError: name 'pritn' is not defined`. В 90 % случаев этой строки достаточно.
- **Выше** — где именно это случилось: номер строки, сам код, иногда стрелочки `^^^^` под виновником.
- **Ещё выше** — цепочка вызовов, как мы туда попали. Пока у нас нет своих функций, она короткая.

Второе правило: **Python часто сам подсказывает решение.** Начиная с версии 3.10 он пишет «Did you mean…?» — «может, вы имели в виду…?». Читайте подсказку, она почти всегда права.

*«Папа, если компьютер знает, что не хватает кавычки, почему он сам её не поставит?» Потому что он не знает, где именно вы её забыли. Угадывать Python отказывается принципиально.*

# 8. Цикл `while`: день сурка

Помните фильм «День сурка»? Герой просыпается, и каждый раз — то же самое утро, та же песня по радио, та же лужа. Пока он не сделает что-то правильно. Цикл в программировании — это день сурка для компьютера: один и тот же блок кода повторяется, **пока** выполняется условие.

До сих пор программы шли строго сверху вниз: команда A, B, C — конец. Цикл позволяет сказать: «Делай X, пока не случится Y». Это основа любой автоматизации. Вместо того чтобы написать `print` сто раз, вы пишете его один раз в цикле. А компьютер не устаёт, не скучает и не просит зарплату.

## Анатомия цикла

`while` переводится как «пока». «ПОКА условие истинно — ДЕЛАЙ». У классического цикла четыре части, и забыть любую — значит, получить проблему:

1. **Подготовка** — до цикла заводим переменную, которая будет им управлять. Счётчик.
2. **Условие** — после `while`. Проверяется **перед каждым** повторением.
3. **Тело** — блок с отступом, то, что повторяется.
4. **Изменение** — внутри тела обязательно делаем что-то, что однажды сделает условие ложным.

````python
# Пример: Считаем от 1 до 5

# 1. Инициализация
counter = 1

# 2. Условие
while counter <= 5:
    # 3. Тело цикла
    print('Цикл выполнился', counter, 'раз(а)')

    # 4. Изменение
    counter += 1 # Эквивалентно counter = counter + 1

# Этот код выполнится только после того, как условие цикла станет False
print("Цикл завершил свою работу.")
````

Каждый проход по телу цикла называется **итерацией**. Здесь их пять. После пятой `counter` стал равен 6, условие `6 <= 5` ложно, и Python вышел из цикла.



*Здесь покоится программист. Прочитал на шампуне: «Намылить. Смыть. Повторить». Условия выхода в инструкции не было. Часть 4 — «изменение» — не забывайте.*

## Бесконечный цикл

А что, если забыть часть 4 — изменение? `counter` навсегда останется единицей, условие `1 <= 5` будет истинно вечно, и цикл никогда не кончится. Это **бесконечный цикл**.

Python его не поймает — с его точки зрения всё законно. Программа просто зависнет, а вентилятор ноутбука начнёт взлетать.

```python
# НЕ ЗАПУСКАТЬ (или запустить и посмотреть, как останавливать)
counter = 1
while counter <= 5:
    print("Я в бесконечном цикле!")
    # забыли counter += 1
```

**Как остановить:**
- В VS Code — кнопка **Interrupt** (квадратик ■) над тетрадкой. Если не помогает — **Restart**, и потом перезапустить нужные ячейки.
- В Jupyter в браузере — то же самое: ■ на панели или меню **Kernel → Interrupt / Restart Kernel**.
- В терминале — `Ctrl+C`.

## Когда `while` незаменим

`while` нужен, когда вы **не знаете заранее**, сколько раз придётся повторять. Классика — проверка ввода. Мы хотим положительный возраст и будем спрашивать, пока человек не введёт нормальное число. Введите сначала `-5`, потом `20`:

````python
age = int(input("Введите ваш возраст: "))

while age <= 0:
    print("Ошибка! Возраст должен быть положительным числом.")
    age = int(input("Пожалуйста, попробуйте снова. Введите ваш возраст: "))

print(f"Спасибо! Ваш возраст ({age}) успешно записан.")
````

Сколько раз спросит? Не знаем. Может, один, может, сорок — смотря насколько пользователь упорный. Ровно для этого и нужен `while`.

### Диалог со стоп-словом

Программа будет болтать с вами, пока вы не скажете «выход». Введите `привет`, потом `выхоД` — с большой буквой в конце, специально:

````python
user_input = "" # Инициализируем пустой строкой

while user_input.lower() != "выход":
    user_input = input("Скажите что-нибудь (или напишите 'выход', чтобы закончить): ")

    # Проверим, чтобы не отвечать на само слово "выход"
    if user_input.lower() != "выход":
        print(f"Вы сказали: '{user_input}'")

print("Программа завершена. До свидания!")
````

«выхоД» сработало, потому что мы сравниваем `user_input.lower()` — сначала приводим к маленьким буквам, потом сравниваем. Опять тот же принцип: сначала почистить, потом сравнить.

Но посмотрите, как коряво вышло: проверка «выход» написана **два раза** — в условии цикла и внутри. И пришлось заранее заводить пустую строку, чтобы цикл вообще начался. Есть способ лучше.

## `break`: стоп-кран

`break` немедленно выходит из цикла, даже если условие ещё истинно. Это стоп-кран в поезде.

А вместе с ним — самая популярная конструкция с `while` в реальном коде: `while True`. Цикл, который **намеренно** бесконечный, а выход из него — только через `break`. Тот же диалог. Введите `как дела`, потом `ВЫХОД`:

````python
# Перепишем наш "диалог" с использованием `while True` и `break`
# `while True` - это самый простой способ создать бесконечный цикл намеренно.
# Выход из такого цикла возможен ТОЛЬКО через `break`. Это очень распространенная идиома.

print("--- Версия диалога с break ---")
while True:
    user_input = input("Скажите что-нибудь (или напишите 'выход', чтобы закончить): ")

    if user_input.lower() == "выход":
        break # Немедленно выходим из цикла

    print(f"Вы сказали: '{user_input}'")

print("Программа завершена. До свидания!")
````

Проверка одна, никаких пустых строк заранее. Читается как жизнь: «Повторяй: спросил, если сказали "выход" — уходим, иначе — ответил».

## `continue`: пропустить ход

`continue` прекращает **текущую** итерацию и сразу переходит к следующей. Код после `continue` в этот раз не выполняется. Как «пропустить ход» в настольной игре.

````python
# Задача: напечатать все нечетные числа от 1 до 10.
number = 0
while number < 10:
    number += 1

    # Если число четное (остаток от деления на 2 равен 0)...
    if number % 2 == 0:
        continue # ...пропускаем оставшуюся часть цикла (print) и идем на следующую итерацию

    print(f"Нечетное число: {number}")
````

Чётные числа дошли до `continue` и не дожили до `print`.

## Моржовый оператор `:=`

Две минуты на штуку, которую вы будете видеть в чужом коде. С Python 3.8 есть оператор `:=` — «присвоить и сразу использовать». Его называют моржовым, потому что `:=` похоже на моржа, повёрнутого набок: глаза и бивни.

Он позволяет записать «спросил и проверил» прямо в условии `while`. Введите `раз`, потом `выход`:

````python
while (command := input("Команда: ")) != "выход":
    print(f"Выполняю: {command}")
print("Готово")
````

`command := input(...)` сначала кладёт введённое в `command`, а потом всё выражение равно этому значению, и его можно сразу сравнить. Скобки здесь нужны ради порядка операций: у моржа самый низкий приоритет, ниже любого сравнения. Без скобок Python сначала вычислит `input(...) != "выход"`, и в `command` попадёт не ваша команда, а `True` или `False`. Код при этом запустится и молча будет делать чушь — худший вид ошибки.

Использовать `:=` не обязательно — `while True` + `break` делает то же самое и понятен всем. Но если увидите моржа — не пугайтесь, он не кусается.

## `else` у цикла

И последняя странность, которая есть почти только в Python. У `while` может быть блок `else`. Он выполняется, если цикл закончился **сам** — условие стало ложным. А если вышли через `break` — `else` не выполняется. Это удобно для поиска: нашли — `break`, не нашли — сработает `else`.

Банкомат даёт три попытки ввести PIN. Введите три неверных: `1111`, `1234`, `0000`:

````python
correct_pin = "7777"
attempt = 1
while attempt <= 3:
    pin = input(f"PIN, попытка {attempt}: ")
    if pin == correct_pin:
        print("PIN верный, деньги выдаём")
        break
    print("Неверно")
    attempt += 1
else:
    print("Три неверных попытки. Карта заблокирована, банкомат её съел.")
````

Все три попытки неверны, `break` не случился — сработал `else`. Перезапустите и введите `7777` — `else` пропадёт.

Честно скажу: многие опытные программисты считают `else` у цикла плохой идеей, потому что слово «иначе» тут сбивает с толку. Правильнее было бы назвать его `nobreak`. Знать — надо, злоупотреблять — нет.

Причина — цикл `while` в программе, которая считала текущую дату. Плеер хранил время как количество дней, прошедших с 1 января 1980 года, и переводил это число в год вот таким циклом. Я переписал оригинальный код на C на Python один в один, только добавил предохранитель, чтобы не повесить нам тетрадку:

````python
def year_from_days(days, year=1980):        # день 1 = 1 января 1980
    steps = 0
    while days > 365:
        steps += 1
        if steps > 10_000:
            return "ЗАВИС 🔁"
        if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):   # високосный год
            if days > 366:
                days -= 366
                year += 1
        else:
            days -= 365
            year += 1
    return year


print("30 декабря 2008:", year_from_days(10592))
print("31 декабря 2008:", year_from_days(10593))
````

`def` — это создание своей функции, разберём в следующей лекции. Сейчас смотрите только на `while`.

## Список — это стеллаж

До сих пор в переменной лежало одно значение: один возраст, одно имя. Для реальных задач этого катастрофически мало. У вас не один клиент, а десять тысяч. Не одна оценка, а целая ведомость.

**Список** (`list`) — это стеллаж с пронумерованными ячейками. В каждую ячейку можно положить что угодно, у каждой — свой номер. Нумерация, как и на улице со строками, — с нуля.

Три свойства списка:
- **Упорядоченный** — элементы лежат в том порядке, в котором вы их положили.
- **Изменяемый** — можно менять, добавлять и выкидывать элементы. Это его главное отличие от строки.
- **Всеядный** — в одном списке могут лежать элементы разных типов. Можно, но обычно не нужно.

Создаётся квадратными скобками, элементы — через запятую:

````python
empty = []
numbers = [1, 2, 3, 4, 5, 3.14]
cities = ["Москва", "Санкт-Петербург", "Казань"]
mixed = ["Привет", 100, True, 3.14, None]

print(empty)
print(numbers)
print(cities)
print(mixed)
print(len(cities), "города")
````

## Индексы и срезы — как у строк

Всё, что мы делали со строками, работает и со списками. Те же индексы, те же отрицательные индексы, те же срезы. Это одна из прекрасных вещей в Python: один раз научились — работает везде.

````python
#            0           1         2        3        4
planets = ["Меркурий", "Венера", "Земля", "Марс", "Юпитер"]
#           -5          -4        -3       -2       -1

print(planets[0])     # первая
print(planets[2])     # мы живём здесь
print(planets[-1])    # последняя в нашем списке

print(planets[0:4])   # индексы 0, 1, 2, 3 — Марс (индекс 3) включён, Юпитер (индекс 4) — нет
print(planets[4:])    # с индекса 4 и до конца
````

В разделе про ошибки я обещал вам `list index out of range`. Выполняю. Планет пять — значит, последняя под номером 5? Попросим `planets[5]`:

````python
print(planets[5])     # ⚠️ упадёт специально
````

Ровно та же ошибка, что и со строками, только теперь `list`. Ещё раз, потому что это важно: в списке из пяти планет **последний индекс — 4**, а не 5. Первая планета — нулевая. Люди, которые пришли из MATLAB, R или Lua, где счёт идёт с единицы, первые две недели страдают. Остальные страдают, когда объясняют это бабушке.



*Первое слово младенца: «массивы индексируются с единицы». Дальше — как видите. Жестоко? Индексы с нуля — тоже.*

## Список можно менять

Строка — фотография, её не изменить. А список — это живой стеллаж: можно залезть в любую ячейку и положить туда что-то другое.

````python
my_pets = ["кот", "собака", "хомяк"]
print(f"Мои питомцы: {my_pets}")

my_pets[2] = "игуана"   # хомяк, покойся с миром
print(f"Мои питомцы (после смерти хомяка): {my_pets}")
````

Исходный список изменился. Никакого «создать новый и переприсвоить», как со строками. Хомяк просто перестал существовать, а на его место заселилась игуана. Жизнь жестока.

## Добавить и выкинуть

У списка куча методов. Главные:

- `.append(x)` — положить в **конец**. Самый частый метод.
- `.insert(i, x)` — вставить на место `i`, остальные сдвинутся вправо.
- `.remove(x)` — найти и выкинуть **первый** элемент, равный `x`. Нет такого — `ValueError`.
- `.pop(i)` — выкинуть элемент с номером `i` и **вернуть** его. Без номера — последний.
- `del список[i]` — просто удалить по номеру.

Идём в магазин:

````python
shopping = ["молоко", "хлеб"]
shopping.append("яйца")
print(shopping)

shopping.insert(0, "масло")    # масло — в первую очередь
print(shopping)

shopping.remove("хлеб")        # передумали
print(shopping)

bought = shopping.pop()        # купили последнее в списке
print(f"Купил: {bought}, осталось: {shopping}")

del shopping[0]                # масло тоже купили
print(shopping)
````

Обратите внимание на `.pop()`: он не просто удаляет, он **отдаёт** вам удалённое. Поэтому его можно положить в переменную `bought`. А `.remove()` и `del` ничего не отдают.

## Вопрос залу: `append` против `extend`

Есть список `[1, 2, 3]`. Мы делаем `extend` с тремя числами, а потом `append` со списком из трёх чисел. *🎤 Спросите: сколько элементов будет в списке? Соберите версии, потом запускайте.*

````python
nums = [1, 2, 3]
nums.extend([4, 5, 6])
print(nums)

nums.append([7, 8, 9])
print(nums)
print(len(nums), "элементов")
````

Семь, а не девять. `extend` берёт **содержимое** списка и раскладывает по одному элементу. А `append` кладёт в конец **ровно одну вещь** — и эта вещь сама является списком. Список внутри списка. Последний элемент — это целая коробка `[7, 8, 9]`, а не три числа.

````python
print(nums[-1])       # последний элемент — список
print(nums[-1][0])    # а в нём — свой нулевой элемент
````

## Сортировка: `.sort()` против `sorted()`

Две похожие вещи, которые ведут себя по-разному, — и это классическая ловушка.

- `список.sort()` сортирует **сам список**, на месте. И ничего не возвращает.
- `sorted(список)` возвращает **новый** отсортированный список. Исходный не трогает.

````python
grades = [4, 5, 3, 5, 2, 4]
new_grades = sorted(grades)
print(grades, new_grades)

grades.sort()
print(grades)

grades.sort(reverse=True)   # по убыванию
print(grades)

oops = grades.sort()
print(oops)                 # вот она, ловушка
````

`None`. Метод `.sort()` отсортировал список и вернул пустоту. Если вы напишете `grades = grades.sort()`, то потеряете свои оценки и получите `None`. Делается это один раз в жизни, зато запоминается навсегда.

## `for` против `while`

Теперь, когда у нас есть списки, познакомимся со вторым циклом.

`while` работает «пока условие истинно». Он универсальный, но для перебора списка неудобен: надо вручную заводить счётчик, следить за ним и не забыть его увеличить. Забыли — бесконечный цикл.

`for` создан ровно для одной задачи: **пройтись по каждому элементу** последовательности.

| `while` | `for` |
| :--- | :--- |
| Работает по **условию** | Работает по **последовательности** |
| «Ешь, ПОКА голоден» | «Съешь КАЖДЫЙ пельмень в тарелке» |
| Счётчик ведёте вы | Перебор ведёт Python |

В девяти случаях из десяти, когда надо перебрать список, `for` проще, безопаснее и короче.

Синтаксис: `for переменная in последовательность:` и блок с отступом. На каждой итерации Python берёт очередной элемент и кладёт его в переменную.

````python
fruits = ["яблоко", "банан", "манго"]

print("--- Наш фруктовый салат ---")
# На каждой итерации в переменную `fruit` будет по очереди попадать "яблоко", "банан", "манго"
for fruit in fruits:
    # `fruit` - это обычная переменная, доступная внутри цикла
    print(f"Нарезаем: {fruit.capitalize()}")

print("Салат готов!")
````

Переменная `fruit` по очереди стала яблоком, бананом и манго. Никаких счётчиков, никаких индексов, никакого риска бесконечного цикла — элементы кончились, цикл кончился.

`for` умеет перебирать не только списки, а вообще всё, что можно перебрать. Например, строку — по одной букве:

````python
vowels = 0
for letter in "Программирование":
    if letter.lower() in "аеёиоуыэюя":
        vowels += 1
print("Гласных:", vowels)
````

Заодно — главный шаблон аналитика, который вы будете писать тысячу раз: **накопитель**. До цикла заводим переменную `vowels = 0`, в цикле — по условию добавляем. После цикла — результат. Сумма, количество, максимум — всё считается этим шаблоном.

## `range()`: повторить N раз

А если нужно просто повторить действие N раз, и никакого списка нет? Для этого есть `range()` — ленивая последовательность чисел: числа в ней не хранятся, а вычисляются, когда их попросят.

- `range(stop)` — от 0 до `stop - 1`.
- `range(start, stop)` — от `start` до `stop - 1`.
- `range(start, stop, step)` — с шагом.

Граница `stop`, как и в срезах, **не включается**.

````python
for i in range(5):
    print("Сообщение номер", i + 1)

print(list(range(2, 10, 3)))

for i in range(3, 0, -1):    # start=3, stop=0 (не включается), step=-1
    print(f"{i}...")
print("Пуск!")
````

`range` не создаёт список в памяти, а выдаёт числа по одному, когда их просят. `range(1_000_000_000)` занимает в памяти столько же, сколько `range(5)`: миллиард чисел нигде не лежит, `range` помнит только начало, конец и шаг. (Подчёркивания внутри числа — просто разделитель разрядов, как пробел в «1 000 000»; Python их не замечает.) Чтобы посмотреть числа глазами, мы превратили его в список через `list()`.

При этом `range` — не одноразовая раздача, а полноценная последовательность: у неё есть длина, индексы и проверка `in`, как у списка. И считается всё это мгновенно, без перебора. Где-то вы услышите, что `range` — «генератор». Это неправда: генераторы в Python — отдельный зверь, до них мы ещё доберёмся, и вот они-то одноразовые.

````python
r = range(0, 1_000_000_000, 5)
print(len(r))           # двести миллионов — посчитано, а не перебрано
print(r[3])             # четвёртое число
print(999_999_995 in r, 7 in r)
````

Двести миллионов чисел, а Python ответил мгновенно и не потратил ни байта на сами числа: длину и «есть ли 7» он вычислил по формуле из начала, конца и шага. Семёрки нет — шаг 5 через неё перешагивает.

## `enumerate()`: номер и значение

Часто нужен не только элемент, но и его номер. Люди, пришедшие из C, делают так:
```python
for i in range(len(tasks)):
    print(i, tasks[i])
```
Работает, но это плохой стиль: длинно, легко ошибиться в индексе, и читается как инструкция к стиральной машине. В Python для этого есть `enumerate()`: он выдаёт пары «номер, элемент».

````python
tasks = ["Сходить на работу/отправить резюме", "Сходить на пару к Воробьеву Д.А.", "Лечь спать пораньше"]

# `enumerate` сразу дает нам и индекс, и значение
for index, task in enumerate(tasks):
    # По умолчанию нумерация с 0, добавим 1 для красоты
    print(f"Задача №{index + 1}: {task}")
````

`for index, task in enumerate(tasks)` — сразу две переменные, Python сам раскладывает пару. Помните `x, y = y, x`? Это та же распаковка. А чтобы не писать `index + 1`, можно попросить нумерацию с единицы: `enumerate(tasks, start=1)`.



*«В массиве моих интересов ты на месте [1]». Она довольна. Она ещё не знает, что это второе место.*

## `break` и `continue` в `for`

Работают точно так же, как в `while`: `break` — выйти из цикла, `continue` — пропустить этот элемент.

````python
numbers = [1, 3, 4, 8, 2, 9, 7]
for num in numbers:
    if num > 5:
        print(f"Первое число больше 5: {num}")
        break

for name in ["Иван", "Мария", "Анна", "Пётр"]:
    if name == "Анна":
        continue      # с Анной сегодня уже здоровались, второй раз — неловко
    print(f"Привет, {name}!")
````

# 10. Кортежи: высечено в камне

Списки мы освоили. Теперь три их родственника, и у каждого своя суперсила: кортеж, множество и словарь. Начнём с самого скучного на вид — и самого недооценённого.

**Кортеж** (`tuple`, по-английски произносят «тьюпл» или «тапл», и в разговоре программисты часто так и говорят — «тапл») — это упорядоченная **неизменяемая** коллекция. Всё как у списка: порядок, индексы, срезы. Но изменить его после создания нельзя. Ни добавить, ни удалить, ни заменить.

> **Список** — это доска, на которой пишут маркером. Дописали, стёрли, исправили.
>
> **Кортеж** — это надпись, вырубленная топором в камне. Что написано, то написано. Хотите другое — берите новый камень.

Зачем нужна такая негибкость? Три причины:

1. **Защита.** Координаты, цвет в RGB, дата рождения — вещи, которые не должны случайно поменяться где-то в недрах вашей программы. Кортеж гарантирует, что их никто не испортит.
2. **Скорость и память.** Python знает, что кортеж не изменится, и хранит его чуть экономнее.
3. **Кортеж может быть ключом словаря.** А список — не может. Почему — увидим через пару разделов.

Создаётся круглыми скобками. А на самом деле — запятыми: скобки можно и не писать.

````python
point = (10, 20, 5)
person = ("Иван", "Петров", 35, 182.5)
another_point = 30, 40          # без скобок — тоже кортеж
print(point, person, another_point)
print(type(another_point))
````

## Ловушка: кортеж из одного элемента

Внимание, вопрос: что такое `(50)`?

````python
not_a_tuple = (50)
real_tuple = (50,)
print(not_a_tuple, type(not_a_tuple))
print(real_tuple, type(real_tuple))
````

`(50)` — это просто число 50 в скобках, как в математике `(2 + 3)`. Кортеж делает **запятая**. Поэтому кортеж из одного элемента пишется с висячей запятой: `(50,)`. Выглядит как опечатка, но это не опечатка.

## Что с кортежем можно, а что нельзя

Всё, что не меняет его: индексы, срезы, длину, `in`, перебор в `for`, методы `.count()` и `.index()`.

````python
data = (10, 20, "a", "b", 30, "c", 10)
print(data[0], data[2:5], len(data))
print("a" in data, "z" in data)
print(data.count(10), data.index("b"))
````

А вот попытка поменять — сразу ошибка. Одной хватит, остальные попытки (`.append`, `del`) заканчиваются так же:

````python
rgb_red = (255, 0, 0)
rgb_red[0] = 100   # ⚠️ упадёт специально
````

`'tuple' object does not support item assignment` — «кортеж не поддерживает присваивание элементов». Камень есть камень.

## Распаковка: главная суперсила кортежа

Вы уже дважды пользовались распаковкой и не знали этого: в `x, y = y, x` и в `for index, task in enumerate(tasks)`. Суть: кортеж можно разложить по переменным одной строкой.

````python
name, surname, age, height = person
print(name, age)

first, *rest = [1, 2, 3, 4, 5]
print(first, rest)

# .split() без аргумента режет строку по пробелам, а + склеивает два списка в один
first, *middle, last = "Анализ данных".split() + ["на", "Python"]
print(first, middle, last)
````

Звёздочка перед именем значит «а сюда — всё, что осталось», и это всегда будет список. Очень удобно, когда надо отделить заголовок от данных: `header, *rows = lines`. Если звёздочки нет, число переменных слева должно точно совпасть с числом элементов справа. Элементов больше — будет `ValueError: too many values to unpack`, меньше — `ValueError: not enough values to unpack`. Python не гадает, кому что досталось.

И последнее: кортеж может быть **ключом словаря**. Например, «координата клетки → что в ней лежит» или «(город, год) → население». К этому вернёмся в разделе про словари.

# 11. Множества: есть или нет

Задача. У вас список хештегов из постов: `["python", "data", "ml", "python", "data"]`. Нужно показать уникальные. Можно написать цикл, завести пустой список и добавлять, если элемента ещё нет… А если хештегов миллион? Для таких задач в Python есть встроенное, быстрое и элегантное решение — **множество** (`set`).

**Множество** — это неупорядоченная коллекция **уникальных** элементов.

- **Уникальных** — дубли выкидываются автоматически. Добавили второй раз то же самое — ничего не произошло.
- **Неупорядоченная** — у элементов нет порядка и нет индексов. `my_set[0]` не существует.

> **Список** — журнал всех входов в здание за день. Иван заходил пять раз — в журнале он пять раз.
>
> **Множество** — список уникальных посетителей за день. Неважно, сколько раз заходил Иван, он там один раз.

````python
tags = ["python", "data", "ml", "python", "data"]
unique_tags = set(tags)
print(unique_tags)
print(len(tags), "->", len(unique_tags))

numbers = {1, 2, 3, 4, 3, 2, 5, 1}   # фигурные скобки без двоеточий
print(numbers)
````

Порядок, в котором Python напечатает множество строк, может меняться от запуска к запуску. Перезапустите ядро и посмотрите. Это не баг — множество просто не обещает никакого порядка. Если нужен порядок — `sorted(unique_tags)` вернёт отсортированный список.

## Зачем множества: скорость

Первая суперсила — удаление дублей — вы уже видели. Вторая — **мгновенная проверка «есть или нет»**.

> Проверка `in` в списке — вы ищете человека в огромной очереди, спрашивая каждого по порядку: «Вы Иван? А вы?»
>
> Проверка `in` в множестве — вы подходите к стойке регистрации, называете имя, и вам сразу говорят: есть или нет.

Как это работает внутри — через хеш-таблицу, о ней чуть позже в словарях. А сейчас — проверим, а не верим на слово. Создадим список и множество из миллиона чисел и поищем последнее. `%timeit` — это «магическая» команда Jupyter, она запускает строку много раз и меряет среднее время. Это не часть Python, в обычном скрипте её не будет. Параметры `-n` и `-r` говорят, сколько запусков подряд и сколько таких серий сделать: без них `%timeit` сам подбирает число повторов, и медленный список мы бы ждали заметно дольше.

````python
large_list = list(range(1_000_000))
large_set = set(large_list)
number_to_find = 999_999

%timeit -n 20 -r 3 number_to_find in large_list
%timeit -n 100000 -r 3 number_to_find in large_set
````

Посмотрите на единицы измерения: у списка — миллисекунды (ms), у множества — наносекунды (ns). Разница — в десятки, а то и сотни тысяч раз. Точные цифры у вас будут свои, зависят от компьютера, — смотрите на порядок. Список честно перебрал миллион элементов. Множество просто сразу посмотрело в нужное место. (Множеству мы дали 100 000 запусков, а не 20: одна проверка длится десятки наносекунд, и на двадцати замерах шум часов был бы больше самого замера.)

Мораль: если вы в цикле много раз проверяете `if x in что-то` — и это «что-то» большое, — сделайте из него множество. Это самая дешёвая оптимизация в мире, и спасает она чаще, чем вы думаете.

## Добавить и выкинуть: философия ошибок

`.add(x)` — добавить один элемент. `.update(коллекция)` — добавить много. Удаляют двумя способами, и это интересно.

````python
online = {"anna", "ivan", "petr"}
online.add("sergey")
online.add("anna")                        # уже есть — ничего не произойдёт
online.update(["olga", "maria", "ivan"])  # ivan не задвоится
print(sorted(online))

online.discard("vladimir")                # нет такого — и ладно
online.remove("petr")                     # есть — удалили
print(sorted(online))
````

`.discard()` молча проглотил Владимира, которого в сети нет. А теперь попросим о том же самом `.remove()`:

````python
online.remove("vladimir")   # ⚠️ упадёт специально
````

Зачем два почти одинаковых метода — «безопасный» `.discard()` и «опасный» `.remove()`? Это не прихоть, это философия.

**`.remove()` — когда элемент обязан там быть.** Если его нет, значит, в программе что-то пошло не так, и громкое падение с `KeyError` — это **хорошо**. Вы кассир в кинотеатре, вам принесли билет на возврат. Билет должен быть в базе проданных. Если его там нет — это поддельный билет, и надо поднять тревогу, а не молча вернуть деньги.

**`.discard()` — когда отсутствие элемента — нормально.** Ваша задача — «чтобы этого здесь не было», а было оно до этого или нет, неважно. Вы убираетесь в комнате и выкидываете старый журнал. Нет журнала — ну и отлично, меньше работы.

Эта идея — «падать, когда случилось то, чего быть не должно» — пройдёт через весь курс. Программа, которая молча проглатывает ошибки, хуже программы, которая падает. Падающую хотя бы видно.

## Математика множеств

Самая красивая часть множеств — операции из школьной теории множеств. Есть две группы сотрудников: разработчики и аналитики. Некоторые совмещают.

````python
devs = {"Иван", "Анна", "Пётр", "Мария"}
analysts = {"Анна", "Сергей", "Мария", "Елена"}

print("Все (объединение |):          ", sorted(devs | analysts))
print("И то, и то (пересечение &):   ", sorted(devs & analysts))
print("Только разработчики (devs - analysts):", sorted(devs - analysts))
print("Только аналитики (analysts - devs):   ", sorted(analysts - devs))
print("Кто не совмещает (симм. разность ^):  ", sorted(devs ^ analysts))
````

Обратите внимание: **`A - B` не равно `B - A`**. «Разработчики, которые не аналитики» и «аналитики, которые не разработчики» — разные люди. Как и в арифметике, `5 - 3` не равно `3 - 5`.

У каждого оператора есть длинное имя-метод: `|` — `.union()`, `&` — `.intersection()`, `-` — `.difference()`, `^` — `.symmetric_difference()`. Делают то же самое, читаются длиннее.

Где это в жизни аналитика? «Какие клиенты купили в январе, но не вернулись в феврале» — это `январь - февраль`. «Кто есть в обеих выгрузках» — `&`. «Кто был хоть где-то» — `|`. Во втором семестре в pandas вы встретите те же операции, только для таблиц. И да — в домашке №2 множества вам пригодятся.

# 12. Словари: телефонная книга

Последняя и самая важная коллекция. Если из сегодняшней лекции вы запомните только одну структуру данных — пусть это будет словарь.

Посмотрите на список `["Иван", "Петров", 35]`. Что значит 35? Возраст? Номер квартиры? Размер обуви? Список хранит значения, но не хранит **смысл**. Нам не хватает подписей.

**Словарь** (`dict`) хранит пары **ключ: значение**.

- **Ключ** — подпись, по которой мы находим значение. Уникален: двух одинаковых ключей в словаре не бывает.
- **Значение** — что угодно: число, строка, список, другой словарь.

> **Список** — плейлист. «Включи трек номер пять». Вы обращаетесь по номеру.
>
> **Словарь** — телефонная книга. Вы не ищете «контакт номер 157». Вы ищете «Иван Петров» и сразу получаете телефон.

Создаётся фигурными скобками с двоеточиями:

````python
student = {
    "name": "Анна",
    "surname": "Петрова",
    "age": 19,
    "course": 2,
    "grades": [5, 4, 5, 5],
}
print(student)
print(student["name"])
print(student["grades"])
print(len(student), "ключей")
````

## Как это работает: хеш-таблица

Почему поиск в словаре — как и в множестве — мгновенный, даже если там миллион записей?

Внутри словаря — **хеш-таблица**. Когда вы кладёте пару `"name": "Анна"`, Python пропускает ключ `"name"` через специальную функцию `hash()`. Она превращает ключ в большое число, и по этому числу Python сразу вычисляет, в какую ячейку памяти положить значение. Когда вы потом просите `student["name"]`, Python снова считает `hash("name")`, получает то же число и идёт **прямо в ту ячейку**. Не перебирает все записи — сразу идёт куда надо. Как если бы в телефонной книге каждая фамилия сама говорила вам номер страницы.

````python
print(hash("name"))
print(hash("name"))   # то же число — пока работает это ядро
print(hash(42))
````

Для строк число будет другим после перезапуска ядра — Python специально перемешивает хеши при каждом запуске, чтобы злоумышленники не могли подобрать ключи, которые попадут в одну ячейку и замедлят сервер. Внутри одного запуска хеш строки всегда одинаковый.

Отсюда же главное правило: **ключом может быть только то, что не меняется** — строки, числа, кортежи. Если бы ключ можно было изменить, его хеш поменялся бы, и Python больше никогда не нашёл бы значение — он искал бы его не в той ячейке. Поэтому список ключом быть не может:

````python
coords = {}
coords[(55.75, 37.62)] = "Москва"   # кортеж — можно
print(coords)
coords[[59.94, 30.31]] = "Питер"    # ⚠️ упадёт специально: список — нельзя
````

`unhashable type: 'list'` — «нехешируемый тип». Вот обещанная причина, по которой кортеж может быть ключом, а список — нет.



*Заповедь, принесённая с горы: «Не знаешь, как решить, — используй хешмап». Половина задач на собеседованиях решается словарём. Вторая половина — множеством.*

## Главная задача словаря: подсчёт

Классика: посчитать, сколько раз встречается каждое слово. Вы будете решать эту задачу всю жизнь в разных обличьях: сколько заказов у каждого клиента, сколько ошибок каждого типа в логе, сколько раз каждый студент опоздал.

Сначала — руками, через `.get()`:

````python
text = "мама мыла раму а рама мыла маму а мама устала"
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
````

`counts.get(word, 0) + 1` — «возьми, сколько было (или ноль, если слова ещё не было), и прибавь один». Эта строчка — шаблон, который надо уметь писать с закрытыми глазами.

А теперь то же самое, как это делают взрослые. В стандартной библиотеке есть модуль `collections`, а в нём — `Counter`, словарь, заточенный под подсчёт:

````python
from collections import Counter

counts = Counter(text.split())
print(counts)
print(counts["мама"])
print(counts["папа"])       # нет такого — 0, а не KeyError
print(counts.most_common(2))
````

Одна строка вместо цикла, плюс `.most_common()` — топ самых частых. Отсутствующий ключ даёт 0 вместо падения.



*Антиплагиат нашёл абзац из Википедии? Удалите абзац из Википедии. А антиплагиат, к слову, начинается ровно с подсчёта слов — `Counter` на стероидах.*

### `defaultdict`: словарь, который не падает

Второй полезный жилец `collections`. Задача: разложить студентов по группам. Руками нужно каждый раз проверять, есть ли уже такая группа в словаре, и если нет — создавать пустой список. `defaultdict` делает это сам: если ключа нет, он создаёт значение по умолчанию.

````python
from collections import defaultdict

students = [("Аня", "БИВТ-26-1"), ("Борис", "БИВТ-26-2"), ("Вика", "БИВТ-26-1"), ("Гоша", "БИВТ-26-1")]

groups = defaultdict(list)       # нет ключа — будет пустой список
for name, group in students:
    groups[group].append(name)
print(dict(groups))
````

`defaultdict(list)` — «если ключа нет, сначала положи туда пустой список». `defaultdict(int)` — «положи ноль» — и это ещё один способ считать.

# 13. `is` против `==`

Маленькая, но коварная тема.

- `==` спрашивает: **равны ли значения?**
- `is` спрашивает: **это один и тот же объект в памяти?** Помните ярлыки? `is` проверяет, приклеены ли два ярлыка к одному и тому же объекту.

````python
a = [5]
b = [5]
c = a
print(a == b, a is b)   # равны, но это два разных списка
print(a == c, a is c)   # c — второй ярлык на тот же список
````

`a` и `b` — два разных списка, в которых случайно лежит одно и то же. `a` и `c` — один и тот же список с двумя ярлыками. Помните коньяк в корзине жены? Вот это `is`.

**Правило на всю жизнь: `is` используется только для сравнения с `None`.** `if result is None:`. Всё остальное — через `==`.

Почему так строго? Потому что с числами и строками `is` ведёт себя непредсказуемо:

````python
x = 256
y = int("256")
print(x is y)

x = 1000
y = int("1000")
print(x is y)           # на Python 3.13 и 3.14 — False, на 3.15 — уже True

x = 2000
y = int("2000")
print(x is y, x == y)
````

256 — один и тот же объект, 1000 и 2000 — разные, хотя значения равны. Дело в том, что CPython (тот Python, который вы поставили) заранее создаёт маленькие целые числа от −5 до 256 и раздаёт их всем желающим, чтобы не плодить копии. А в свежем Python 3.15 этот запас расширили до 1024 — и там на 1000 вы получите уже `True`. То есть один и тот же код с `is` даёт разные ответы в зависимости от версии Python. Это деталь реализации, на неё нельзя полагаться — поэтому `is` только для `None`.

Python, кстати, сам предупредит, если вы напишете `x is 2000` прямо с числом: `SyntaxWarning: "is" with 'int' literal. Did you mean "=="?`.

## 🧨 Экстра: сравнения с подвохом

> Можно пропустить — тогда читаете дома.

Интерактив «угадай вывод». *🎤 Показывайте по одной строке, пусть зал голосует до запуска.*

````python
s1 = "hello"
s2 = "".join(["hel", "lo"])
print(s1 == s2, s1 is s2)
````

Строки равны, но это разные объекты: одну Python создал при чтении кода, вторую — склеил на ходу. Именно поэтому строки сравнивают только через `==`.

Дальше — **NaN**, «не число» (Not a Number). Это специальное значение float, которое получается из бессмысленных операций вроде бесконечность минус бесконечность. Во втором семестре в pandas оно будет означать «пропуск в данных», и вы будете встречать его каждый день.

````python
nan = float("nan")
print(nan == nan)
print(nan in [nan])
````

NaN **не равен сам себе**. Это по стандарту: «неизвестно что» не может быть равно другому «неизвестно что». Поэтому `x == float("nan")` — не способ проверить пропуск, для этого есть `math.isnan(x)`. А `nan in [nan]` — `True`, потому что `in` сначала проверяет `is`, а уже потом `==`. Это тот самый объект — значит, нашёлся.

И финальный босс:

````python
print(False == False in [False])
print((False == False) in [False])
````

Первое — `True`, второе — `False`. Как?! Помните цепочки сравнений `3 <= x < 6`? Python считает цепочкой **любые** операторы сравнения, а `in` и `==` — тоже операторы сравнения. Поэтому `False == False in [False]` — это `(False == False) and (False in [False])`, то есть `True and True`. А со скобками — `True in [False]`, это `False`.

Мораль: в сложных условиях ставьте скобки. Даже если вы точно знаете приоритеты. Особенно если вы точно знаете приоритеты.

Компания друзей съездила в поездку. Всё это время они записывали траты в общий чат. Теперь надо понять, кто кому сколько должен. Знакомая боль?

Ваш вариант — это журнал трат: многострочная строка в тройных кавычках, примерно такого вида:

```text
vasya заплатил 1200 за такси, на всех
```
Видите, сколько тут грязи? Один и тот же человек записан тремя способами — поэтому в блоке варианта есть ещё и таблица псевдонимов (словарь). Суммы — то с пробелом, то с «руб», то с «р». Программа ничего не спрашивает у пользователя — она берёт журнал из блока варианта и выдаёт:
1. **Баланс каждого**: сколько он заплатил минус сколько должен был. Плюс — ему должны, минус — должен он. (Словарь.)
2. **Кто ни разу не платил** и **кто участвовал во всех тратах**. (Множества и операции над ними.)
3. **Список переводов** «кто → кому → сколько», чтобы все рассчитались. Отсортированный, чтобы было удобно читать. (Кортежи и сортировка.)
